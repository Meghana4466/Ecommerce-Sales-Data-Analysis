```python
print('Total Sales:',round(df['Sales'].sum(), 2))
print('Total Profit:',round(df['Profit'].sum(), 2))
print('Total Quantity:',df['Quantity'].sum())
print('Average Sales:',round(df['Sales'].mean(), 2))
```

```python
category_sales=df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
print(category_sales)
```

```python
category_profit=df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
print(category_profit)
```

```python
subcategory_sales=df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)
print(subcategory_sales)
```

```python
region_sales=df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
print(region_sales)
```

```python
segment_sales=df.groupby('Segment')['Sales'].sum().sort_values(ascending=False)
print(segment_sales)
```

```python
import numpy as np
sales=df['Sales'].to_numpy()
print('Average Sales',round(np.mean(sales),2))
print('Minimum Sales',round(np.min(sales),2))
print('Maximum Sales',round(np.max(sales),2))
print('Standard Deviation',round(np.std(sales),2))
```

```python
import matplotlib.pyplot as plt

category_sales.plot(kind='bar')
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_profit.plot(kind='bar')

plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_sales.plot(kind='bar')

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
segment_sales.plot(kind='bar')

plt.title('Total Sales by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

sns.histplot(data=df, x='Sales', bins=30, kde=True)

plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.histplot(data=df, x='Sales', bins=30)
plt.xlim(0, 1000)
plt.title('Sales Distribution (0–1000)')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Category', y='Profit')
plt.title('Profit Distribution by Category')
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    df[['Sales', 'Quantity', 'Discount', 'Profit']].corr(),
    annot=True,
    cmap='coolwarm',
    vmin=-1,
    vmax=1
)

plt.title('Sales and Profit Correlation')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x='Discount', y='Profit', alpha=0.5)

plt.title('Discount vs Profit')
plt.grid()
plt.show()
```

```python
category_sales.plot(kind='bar')
plt.title('Total Sales by Category')
plt.tight_layout()
plt.savefig('sales_by_category.png', dpi=300)
plt.show()
```

```python
print("BUSINESS INSIGHTS")
print("1. Highest sales category:", category_sales.idxmax())
print("2. Highest profit category:", category_profit.idxmax())
print("3. Highest sales region:", region_sales.idxmax())
print("4. Highest sales customer segment:", segment_sales.idxmax())
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Sales Share by Category')
plt.savefig('sales_pie_chart.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.4}
)

plt.title('Sales Share by Category - Donut Chart')
plt.savefig('sales_donut_chart.png', dpi=300)
plt.show()
```
